### 1. Buatlah simulasi deret MA (1) dengan panjang pengamatan 20 dengan nilai parameter $\theta=0.1, \theta=0.3, \theta=0.7, \theta=1.0, \theta=1.5, \theta=2.0$

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima_process import ArmaProcess
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

import warnings
warnings.filterwarnings("ignore")
```

A. Simulasi deret MA (1) kita lakukan secara manual berdasarkan persamaan proses MA (1) dengan $\theta=0.1$ dengan panjang pengamatan sebanyak 20

```python
np.random.seed(120)
theta=0.1
et=[np.random.normal(0,1)]
ma1_m1=[]
for i in range(20):
    et.append(np.random.normal(0,1))
    ma1_m1.append(et[i+1]-theta*et[i])
```

```python
plt.figure(figsize=(10,5))
plt.plot(ma1_m1)
plt.xlabel('Waktu'), plt.ylabel(r'$Y_t$')
plt.show()
```

```python
plt.figure(figsize=(6,5))
pd.plotting.lag_plot(pd.DataFrame(ma1_m1), lag=1, c='k')
plt.xlabel(r'$Y_{t-1}$'),plt.ylabel(r'$Y_{t}$')
plt.show()
```

```python
plt.figure(figsize=(6,5))
pd.plotting.lag_plot(pd.DataFrame(ma1_m1), lag=2, c='k')
plt.xlabel(r'$Y_{t-2}$'),plt.ylabel(r'$Y_{t}$')
plt.show()
```

```python
plot_acf(np.array(ma1_m1), zero=False)
plt.show()
```

Berdasarkan plot pencar dan plot ACF, deret MA(1) dengan \(\theta=0.1\) menunjukkan autokorelasi negatif yang relatif lemah pada lag 1. Secara teoritis, nilai autokorelasi lag 1 adalah \(\rho_1=-\theta/(1+\theta^2)\), sehingga untuk \(\theta=0.1\) diperoleh nilai sekitar \(-0.099\). Sementara itu, autokorelasi pada lag 2 dan lag berikutnya secara teoritis bernilai nol. Hasil simulasi menunjukkan pola yang sejalan dengan teori, meskipun nilai ACF sampel tidak sama persis dengan nilai teoritis karena jumlah pengamatan yang relatif sedikit, yaitu 20 pengamatan.
